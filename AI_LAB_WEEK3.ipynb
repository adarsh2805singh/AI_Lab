```python
def display_puzzle(puzzle):
    print("\nPuzzle:")
    for row in puzzle:
        print(row)


def misplaced_tiles(current, goal):
    count = 0

    for i in range(3):
        for j in range(3):
            # 0 represents the blank space
            if current[i][j] != 0:
                if current[i][j] != goal[i][j]:
                    count += 1

    return count


# Goal state
goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

# Current state
current = [
    [1, 2, 3],
    [4, 6, 5],
    [7, 8, 0]
]

print("8-Puzzle Problem")

print("\nInitial State:")
display_puzzle(current)

print("\nGoal State:")
display_puzzle(goal)

# Calculate heuristic value
h = misplaced_tiles(current, goal)

print("\nMisplaced Tiles =", h)
```

```python
def display_puzzle(puzzle):
    print("\nPuzzle:")
    for row in puzzle:
        print(row)


def manhattan_distance(current, goal):
    distance = 0

    for i in range(3):
        for j in range(3):

            # Ignore the blank space
            if current[i][j] == 0:
                continue

            tile = current[i][j]

            # Find the position of this tile in the goal state
            for x in range(3):
                for y in range(3):
                    if goal[x][y] == tile:

                        # Manhattan distance
                        distance += abs(i - x) + abs(j - y)

    return distance


# Goal state
goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

# Current state
current = [
    [1, 2, 3],
    [4, 6, 5],
    [7, 8, 0]
]

print("8-Puzzle Problem")

print("\nInitial State:")
display_puzzle(current)

print("\nGoal State:")
display_puzzle(goal)

# Calculate Manhattan distance
h = manhattan_distance(current, goal)

print("\nManhattan Distance =", h)
```
